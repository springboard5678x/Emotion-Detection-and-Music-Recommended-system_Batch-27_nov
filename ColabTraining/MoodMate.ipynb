```python
import pandas as pd

music = pd.read_csv('Music_Info[1].csv')
music.head()
```

```python
print(music.columns.tolist())
```

```python
music = music[['name','artist','tags','genre','valence','energy','danceability','acousticness','tempo','spotify_preview_url', 'spotify_id']]
music.head()
```

```python
music['tags'] = music['tags'].fillna('').str.lower()
music['tags_list'] = music['tags'].apply(lambda x: [t.strip() for t in x.split(',')])
```

```python
music['tags_list'].head()
```

```python
mood_tags = [
    'happy','sad','angry','calm','relax','relaxing','chill','chillout',
    'romantic','love','emotional','melancholy','melancholic','energetic',
    'upbeat','dance','danceable','soft','soothing','dark','bright','mellow',
    'aggressive','heavy','party','uplifting','beautiful','acoustic',

    # add these:
    'rage','furious','intense','chaotic','eerie','haunting','disturbing',
    'heartbreak','lonely','cry','depressed','pain','grief'
]
```

```python
def filter_mood_tags(taglist):
    return [t for t in taglist if t in mood_tags]

music['mood_tags'] = music['tags_list'].apply(filter_mood_tags)
```

```python
music[['tags_list','mood_tags']].head()
```

```python
music['mood_text'] = music['mood_tags'].apply(lambda x: ' '.join(x))
music['mood_text'].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(music['mood_text'])
```

```python
tfidf_matrix.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

audio_features = music[['valence','energy','danceability','acousticness','tempo']].fillna(0)

scaler = MinMaxScaler()
audio_norm = scaler.fit_transform(audio_features)
```

```python
from scipy.sparse import hstack

final_features = hstack([tfidf_matrix, audio_norm])
```

```python
emotion_to_mood = {
    "happy": ["happy","uplifting","dance","upbeat"],
    "sad": ["sad","melancholic","emotional"],
    "angry": ["angry","aggressive","heavy"],
    "calm": ["calm","relax","soothing"],
    "love": ["love","romantic","soft"],
    "neutral": ["chill","mellow"],
    "disgust": ["dark","alternative"]
}

emotion_audio_profile = {
    "happy":   [0.8, 0.7, 0.8, 0.2, 0.7],
    "sad":     [0.2, 0.3, 0.2, 0.8, 0.3],
    "angry":   [0.3, 0.9, 0.6, 0.1, 0.8],
    "calm":    [0.6, 0.2, 0.3, 0.9, 0.2],
    "love":    [0.7, 0.4, 0.4, 0.6, 0.4],
    "neutral": [0.5, 0.5, 0.5, 0.5, 0.5],
    "disgust": [0.2, 0.6, 0.3, 0.1, 0.5]
}
```

```python
final_features.shape
```

```python
import numpy as np
import scipy

scipy.sparse.save_npz('music_features.npz', final_features)
music.to_csv('music_meta.csv', index=False)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from scipy.sparse import hstack
import numpy as np

def recommend_music(emotion, top_n=5):
    # 1. Mood text vector
    mood_words = emotion_to_mood.get(emotion, [])
    mood_query = " ".join(mood_words)
    mood_vec = tfidf.transform([mood_query])

    # 2. Emotion-based audio vector
    audio_vec = np.array(emotion_audio_profile[emotion]).reshape(1, -1)

    # 3. Combine text + audio
    emotion_feature = hstack([mood_vec, audio_vec])

    # 4. Similarity
    similarity = cosine_similarity(emotion_feature, final_features)

    # 5. Get top results
    top_indices = similarity[0].argsort()[-20:]
    np.random.shuffle(top_indices)
    top_indices = top_indices[:top_n]

    return music.iloc[top_indices][
        ['name', 'artist', 'spotify_id', 'spotify_preview_url']
    ]
```

```python
recommendations = recommend_music("happy", top_n=5)
recommendations
```

```python
recommendations[['name', 'spotify_preview_url']].head(10)
```

```python
from IPython.display import HTML, display

def show_recommendation(row, detected_emotion):
    song_name = row['name']
    artist = row['artist']
    spotify_id = row['spotify_id']

    print(f"😊 Detected Mood : {detected_emotion}")
    print(f"🎵 Song         : {song_name}")
    print(f"🎤 Artist       : {artist}")

    spotify_url = f"https://open.spotify.com/track/{spotify_id}"
    display(HTML(f'<a href="{spotify_url}" target="_blank">▶️ Open in Spotify</a>'))
```

```python
detected_emotion = "happy"   # from image/text model

for i in range(len(recommendations)):
    print(f"\nRecommendation {i+1}")
    show_recommendation(recommendations.iloc[i], detected_emotion)
```
